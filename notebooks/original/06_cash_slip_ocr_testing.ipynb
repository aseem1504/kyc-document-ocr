```python
def cashslip_fnc(cashslip_url,words):
    print(cashslip_url)
    url = FASTAG_API_BASE+"/escalations/docs/cashslip"

    payload = json.dumps({ "cashSlipURL":cashslip_url, "wordsToFind": words })

    headers = {'Content-Type': 'application/json'}
    
    response = requests.request("POST", url, headers=headers, data=payload)

    dict_vehicle={'cashslip_url':cashslip_url}
    
    try:
    
        for i in json.loads(response.text)['data']['detections']:
    
            dict_vehicle[i['value']]=i['success']
    except : 
        dict_vehicle['no_data']='no_data'
    
    print(dict_vehicle)
        
    return dict_vehicle
```

```python
audit_data['fnc_revert']=audit_data.apply(lambda row : cashslip_fnc(row['Image URL'],['Vehicle','High','Jou']),axis=1)  

```

```python
ocr_data=pd.json_normalize(audit_data.fnc_revert)
```

```python
final=pd.merge(audit_data,ocr_data,left_on='Image URL',right_on='cashslip_url')
```
